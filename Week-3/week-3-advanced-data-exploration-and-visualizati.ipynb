## Discussion and Interpretation

The advanced exploratory analysis provides a deeper understanding of factors associated with campus placement outcomes. The visualizations consistently show that placed students tend to perform better across several academic and employability measures. Rather than relying on a single variable, the analysis demonstrates the importance of examining the student's overall profile.

The box plots provide evidence that the distributions of SSC, HSC, degree, employability test, and MBA scores differ between placement groups. The correlation heatmap further shows relationships among the academic variables, suggesting that students who perform well academically often demonstrate relatively strong performance across multiple stages of education.

Work experience emerged as an important categorical factor. Students with previous work experience showed a higher placement rate, and the chi-square test confirmed that the association between work experience and placement status was statistically significant. This suggests that practical exposure may be relevant to placement outcomes, although the analysis cannot establish a causal relationship.

The SSC hypothesis test also produced a statistically significant result. This means that the observed difference in SSC scores between placed and non-placed students is unlikely to be explained by random sampling variation alone under the assumptions of the test.

Specialisation analysis showed differences between Mkt&Fin and Mkt&HR placement outcomes. However, these differences should be interpreted in the context of employer demand, student preferences, academic performance, and other characteristics.

Overall, the results support a multidimensional view of campus placement. Academic achievement, employability performance, practical experience, and specialisation may all contribute to differences in placement outcomes. Statistical significance should not be confused with causation, and further analysis using larger and more representative datasets would be required to establish predictive or causal relationships.

## Recommendations

Based on the findings, the following recommendations can be considered:

1. Students should maintain consistent academic performance across SSC, HSC, degree, and MBA levels.

2. Students should participate in internships, projects, and other activities that provide practical exposure and work-related experience.

3. Employability-test preparation should be included in placement preparation because employability performance differs across placement groups.

4. Career guidance can help students understand how different specialisations may align with employer requirements.

5. Placement preparation should not focus exclusively on academic marks. Communication skills, practical experience, technical skills, aptitude preparation, and interview readiness should also be developed.

6. Institutions can use historical placement data to identify groups that may benefit from targeted training and career-support programs.

7. Future analysis should use larger datasets from multiple institutions and recruitment cycles to improve the reliability and generalisability of the findings.

## Limitations

The analysis has several limitations. First, the dataset contains only 215 student records, which limits the generalisability of the findings. Second, the data represents a particular campus-placement context and may not represent students from other institutions, regions, or recruitment periods.

The analysis identifies associations rather than causal relationships. Statistical significance does not prove that a particular variable directly causes placement success. In addition, salary information is available only for placed students, so salary comparisons between placed and non-placed students are not appropriate.

Some categorical groups may also contain relatively fewer observations, which can make percentage-based comparisons less stable. Future work could address these limitations by using larger, more diverse datasets and applying multivariate statistical or machine-learning techniques.
